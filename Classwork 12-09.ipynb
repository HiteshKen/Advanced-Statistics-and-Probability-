#### 1. Load the time series dataset using Pandas

```python
import pandas as pd
data = pd.read_csv("sales_timeseries.csv")
print(data)
```

### 🔹 Working
- `pandas` is imported to work with the dataset.
- `read_csv()` reads the CSV file.
- The dataset is stored in the variable `data` and displayed.

#### 2. Convert the date column into a proper DateTime format. 

```python
data["Date"] = pd.to_datetime(data["Date"])
print(data["Date"])
```

### 🔹 Working
- `pd.to_datetime()` converts the Date column into DateTime format.
- This makes the date suitable for time series analysis.

#### 3. Set the date column as the index. 

```python
data["Date"] = pd.to_datetime(data["Date"])
data.set_index("Date", inplace=True)

print(data.head())
```

### 🔹 Working
- `set_index()` makes the Date column the index.
- This helps in performing time-based operations easily.

#### 4. Display the first 10 observations of the dataset.


```python
print(data.head(10))
```

### 🔹 Working
- `head(10)` displays the first 10 observations of the dataset.

#### 5. Plot the time series data using a line graph. 

```python
import matplotlib.pyplot as plt

plt.plot(data.index, data["Revenue"])
plt.title("Sales Over Time")
plt.xlabel("Date")
plt.ylabel("Revenue")
plt.show()
```

### 🔹 Working
- The line graph shows how Revenue changes with time.
- Date is represented on the X-axis.
- Revenue is represented on the Y-axis.

#### 6. Calculate the mean, minimum, and maximum sales values.

```python
print("Mean:", data["Revenue"].mean())
print("Minimum:", data["Revenue"].min())
print("Maximum:", data["Revenue"].max())
```

### 🔹 Working
- `mean()` calculates the average sales.
- `min()` finds the minimum sales value.
- `max()` finds the maximum sales value.

#### 7. Plot the original sales data along with its moving average. 

```python
data["Moving Average"] = data["Revenue"].rolling(3).mean()

plt.plot(data.index, data["Revenue"], label="Sales")
plt.plot(data.index, data["Moving Average"], label="Moving Average")
plt.title("Sales with Moving Average")
plt.xlabel("Date")
plt.ylabel("Revenue")
plt.legend()
plt.show()
```

### 🔹 Working
- `rolling(3).mean()` calculates the average of every 3 consecutive values.
- The moving average smooths the sales data.
- It helps to understand the overall trend more clearly.

#### 8. Identify whether the dataset shows an increasing, decreasing, or approximately stable trend

```python
first = data["Revenue"].iloc[0]
last = data["Revenue"].iloc[-1]

if last > first:
    print("Trend: Increasing")
elif last < first:
    print("Trend: Decreasing")
else:
    print("Trend: Stable")
```

### 🔹 Working
- The first and last Revenue values are compared.
- If the last value is greater, the trend is increasing.
- If the last value is smaller, the trend is decreasing.
- Otherwise, the trend is stable.

#### 9. Predict the sales value for the next time period using a simple method such as the previous value or moving average.

```python
prediction = data["Revenue"].tail(3).mean()
print("Predicted next sales:", prediction)
```
